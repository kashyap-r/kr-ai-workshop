```python
import asyncio


async def download_file(file):
    print(f"Starting {file}")

    await asyncio.sleep(2)

    print(f"Finished {file}")


async def main():

    await asyncio.gather(
        download_file("a.pdf"),
        download_file("b.pdf"),
        download_file("c.pdf")
    )


asyncio.run(main())
```

```python
import asyncio


async def search_vector_db():
    await asyncio.sleep(1)
    return "vector results"


async def query_hr_system():
    await asyncio.sleep(1.5)
    return "HR results"


async def fetch_sweden_policy():
    await asyncio.sleep(1)
    return "Sweden policy"


async def main():

    vector_results, hr_results, sweden_policy = await asyncio.gather(
        search_vector_db(),
        query_hr_system(),
        fetch_sweden_policy()
    )

    print(vector_results)
    print(hr_results)
    print(sweden_policy)


asyncio.run(main())
```
