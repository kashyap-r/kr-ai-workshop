```python
from multiprocessing import Process


def calculate(n):
    print (f"Starting {n}")

    total = 0
    for i in range(50_000_000):
        total += i

    print (f"Finished {n}")

if __name__ == "__main__":
    processes = []

    for n in range(4):
        process = Process(target =calculate, args=(n,))
        process.start()
        processes.append(process)

    for process in processes:
        process.join()




```
