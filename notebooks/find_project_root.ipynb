```python
from pathlib import Path
from dotenv import load_dotenv
import os

# tries to find the .env file in the current working directory and loads it if found. 
# It then retrieves the GROQ_API_KEY from the environment variables and prints whether the API key was successfully loaded.
# project_root = Path.cwd()
# I need to naviagate to its parent folder and then look for the .env folder
project_root = Path.cwd().parent

#  Alternatively, you can use the following code to find the .env file in the parent directory of the current working directory:
# env_file = project_root.parent / ".env"
# OR 
# # Moves down into a subfolder named "config"
# env_file = project_root / "config" / ".env"
# OR 
# Pure shortcut syntax for the parent directory
# env_file = project_root / ".." / ".env"

print("Current working directory:", project_root)

env_file = project_root / "llm-client-demo" / ".env"
if not env_file.exists():
    print("No .env found here")
else:
    print(".env found")

load_dotenv(env_file)
api_key = os.getenv("GROQ_API_KEY")
print("API key loaded:", bool(api_key))
```
