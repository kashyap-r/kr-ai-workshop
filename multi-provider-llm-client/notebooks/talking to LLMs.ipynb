```python
# I need to find the list of active models available in the groq API and print their details. 
# I will use the Groq Python client to list the models and print their ID, owner, context window, and active status.

import os
from pathlib import Path
from dotenv import load_dotenv
from groq import Groq

# 1. Get the directory where THIS notebook is located
notebook_dir = Path().resolve()

# 2. Load the .env file from the parent of notebook directory
env_file = notebook_dir.parent / "llm-client-demo" / ".env"

# 3. Load the environment variables from the .env file 
load_dotenv(dotenv_path=env_file)

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise RuntimeError("GROQ_API_KEY is not set")

client = Groq(api_key=api_key)

models = client.models.list()


for model in models.data:
    print(f"ID:       {model.id}")
    print(f"Owner:    {getattr(model, 'owned_by', 'N/A')}")
    print(f"Context:  {getattr(model, 'context_window', 'N/A')}")
    print(f"Active:   {getattr(model, 'active', 'N/A')}")
    print("-" * 60)
```

Full Structure of the ChatCompletion Object

The <class 'groq.types.chat.chat_completion.ChatCompletion'> object is a structured data container (built using Pydantic). It conforms to the OpenAI standard API response format but includes special Groq-specific extras.The main parent response object holds the metadata for the API call, tracking IDs, models used, and crucial metric data.

The compelte structure of the ChatCompletion Object is given below


response (ChatCompletion Object)
├── .id                 -> Unique ID string for this specific API request
├── .object             -> Always a string value: "chat.completion"
├── .created            -> Unix timestamp integer (e.g., 1718223041)
├── .model              -> The exact model string that responded (e.g., "openai/gpt-oss-20b")
├── .system_fingerprint -> Backend configuration ID tracking current server builds
│
├── .choices (List)     -> The actual responses (Usually containing 1 element)
│    └── Choice Object
│         ├── .index            -> Integer index of the choice (starts at 0)
│         ├── .finish_reason    -> String detailing why it stopped ('stop', 'length', 'tool_calls')
│         ├── .logprobs         -> Token probability metrics if requested (None by default)
│         └── .message          -> ChatCompletionMessage Object
│              ├── .role        -> String ("assistant")
│              ├── .content     -> String (The text answer)
│              └── .reasoning   -> String (Groq-specific internal thinking process)
│
└── .usage (Usage Object)-> Tracking metrics for speed and resource limits
     ├── .prompt_tokens     -> Tokens read from your input prompt
     ├── .completion_tokens -> Tokens written out by the model response
     └── .total_tokens      -> Sum total of prompt + completion tokens
